```python
# Passo 1: Import de Bibliotecas
import hashlib
import ipywidgets as widgets
from IPython.display import display, clear_output
from google.colab import files
```

```python
# Passo 2: Upload de Evidencia
print("Selecione o arquivo da evidência digital:")

uploaded = files.upload()

nome_arquivo = next(iter(uploaded))

print(f"\nArquivo recebido: {nome_arquivo}")
```

```python
# Passo 3: Cálculo do Hash
def calcular_sha256(nome_arquivo):

    sha256 = hashlib.sha256()

    with open(nome_arquivo, "rb") as arquivo:

        for bloco in iter(lambda: arquivo.read(4096), b""):
            sha256.update(bloco)

    return sha256.hexdigest()


hash_calculado = calcular_sha256(nome_arquivo)
```

```python
# Passo 4: Imprimindo o hash do Arquivo
print("\nSHA-256 calculado pelo sistema:")
print(hash_calculado)
```

```python
# Passo 5: Comparando com o hash gerado por outro colega de sala
print("\nInforme o SHA-256 registrado no momento da aquisição:")

campo_hash = widgets.Text(
    placeholder="Cole aqui o hash de referência",
    description="SHA-256:",
    layout=widgets.Layout(width="800px")
)

botao = widgets.Button(
    description="Verificar integridade",
    button_style="primary"
)

resultado = widgets.Output()

display(campo_hash)
display(botao)
display(resultado)
```

```python
# Passo 6: Verificando a integridade
def verificar_integridade(b):

    with resultado:

        clear_output()

        hash_informado = campo_hash.value.strip().lower()

        if not hash_informado:

            print("⚠ Informe o hash de referência.")

            return

        if hash_informado == hash_calculado.lower():

            print("✓ EVIDÊNCIA ÍNTEGRA")
            print()
            print("O SHA-256 calculado para o arquivo")
            print("é igual ao hash registrado na aquisição.")

        else:
            print("✗ EVIDÊNCIA NÃO ÍNTEGRA")
            print()
            print("O SHA-256 calculado para o arquivo")
            print("é diferente do hash registrado na aquisição.")


botao.on_click(verificar_integridade)
```
